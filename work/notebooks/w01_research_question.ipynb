## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import numpy as np

url = 'https://raw.githubusercontent.com/coderabdulrehman1/FlyRankAI/refs/heads/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url)
print( df.columns.tolist())
df.head()


# Total inventory count
total_pages = len(df)
total_clicks_90d = df['clicks_90d'].sum()


print("\n")

# Sort pages to find how many top pages drive the majority of traffic
df_sorted = df.sort_values(by='clicks_90d', ascending=False).copy()
df_sorted['cumulative_clicks'] = df_sorted['clicks_90d'].cumsum()
df_sorted['cumulative_pct'] = (df_sorted['cumulative_clicks'] / total_clicks_90d) * 100

# Find how many pages it takes to hit 80% of total traffic
top_pages_80 = df_sorted[df_sorted['cumulative_pct'] <= 80]
count_top_80 = len(top_pages_80) + 1  # Add 1 to include the crossover page
pct_top_inventory = (count_top_80 / total_pages) * 100

# Count 'Dead Weight' (pages with zero sessions or clicks in the last 30 days)
dead_weight_df = df[df['sessions_last_30d'] == 0]
count_dead_weight = len(dead_weight_df)
pct_dead_weight = (count_dead_weight / total_pages) * 100

print(f"Number 1: {pct_top_inventory:.1f}% of our inventory drives 80% of total clicks.")
print(f"Number 2: 80.0% (concentrated traffic footprint)")
print(f"Number 3: {pct_dead_weight:.1f}% of our pages ({count_dead_weight} pages) are completely dead (0 sessions last 30d).")



print("\n")

# Identify high-exposure content that is declining
# We look at pages with negative trends
declining_df = df[df['trend_pct'] < 0]
total_declining_impressions = declining_df['impressions_90d'].sum()

# Average days since last update for these declining pages
avg_days_stale = declining_df['days_since_last_update'].mean()

# Average decay percentage
avg_decay_pct = declining_df['trend_pct'].mean()

print(f"Number 1: {total_declining_impressions:,} total impressions are locked up in downward-trending assets.")
print(f"Number 2: {avg_days_stale:.1f} days is the average time since these decaying pages were last updated.")
print(f"Number 3: {abs(avg_decay_pct):.1f}% average traffic loss (decay rate) across these stale assets.")



print("\n")

if 'ai_sessions_90d' in df.columns and df['ai_sessions_90d'].sum() > 0:
    total_ai_sessions = df['ai_sessions_90d'].sum()

    # Check if a specific AI model or provider dominates AI sessions
    # Group by model_used to see where AI search traffic is sticking
    ai_model_summary = df.groupby('model_used').agg(
        total_ai_traffic=('ai_sessions_90d', 'sum'),
        avg_scroll_rate=('scroll_rate', 'mean')
    ).reset_index()

    # Sort to find the top AI-performing structure layout
    ai_model_summary = ai_model_summary.sort_values(by='total_ai_traffic', ascending=False)

    if not ai_model_summary.empty:
        top_model = ai_model_summary.iloc[0]['model_used']
        top_model_traffic_pct = (ai_model_summary.iloc[0]['total_ai_traffic'] / total_ai_sessions) * 100
        top_model_scroll = ai_model_summary.iloc[0]['avg_scroll_rate']

        # Express scroll rate cleanly if it's stored as a decimal (0.45 -> 45%)
        if top_model_scroll <= 1.0:
            top_model_scroll *= 100

        print(f"Top performing AI configuration archetype detected: '{top_model}'")
        print(f"Number 1: {top_model_traffic_pct:.1f}% of our total AI engine search sessions land on this archetype.")
        print(f"Number 2: {top_model_scroll:.1f}% average scroll rate (uncovering user engagement drop-offs).")
    else:
        print("Columns exist, but no AI model-driven grouping data found.")
else:
    print("Note: 'ai_sessions_90d' is currently zero or empty in the dataset. Skipping AI metrics breakdown.")

print("="*60)

# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```
